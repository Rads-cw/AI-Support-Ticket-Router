```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(
    df["ticket"],
    df["category"],
    test_size=0.25,
    random_state=42,
    stratify=df["category"]
)

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2)
)

X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

model = LogisticRegression(
    max_iter=1000
)

model.fit(X_train_vectorized, y_train)

print("Training complete!")
print("Training tickets:", len(X_train))
print("Testing tickets:", len(X_test))
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
def get_priority(ticket_text, category):
    text = ticket_text.lower()

    account_high = [
        "locked",
        "account locked",
        "cannot access",
        "can't access",
        "unable to access",
        "cannot log in",
        "can't log in",
        "unable to log in",
        "cannot sign in",
        "can't sign in",
        "hacked",
        "compromised",
        "unauthorized access",
        "someone accessed",
        "security breach",
        "verification code not received",
        "verification code not arriving",
        "account disabled",
        "account blocked"
    ]

    technical_high = [
        "crash",
        "crashing",
        "crashed",
        "blank screen",
        "screen goes blank",
        "black screen",
        "freezes",
        "freezing",
        "frozen",
        "not responding",
        "unresponsive",
        "cannot open",
        "can't open",
        "won't open",
        "will not open",
        "cannot load",
        "won't load",
        "stopped working",
        "not working",
        "server down",
        "offline",
        "cannot save",
        "save failed",
        "cannot upload",
        "upload failed",
        "lost data",
        "data lost",
        "deleted files",
        "missing files"
    ]

    billing_high = [
        "charged twice",
        "double charged",
        "duplicate charge",
        "charged again",
        "still charged",
        "charged after",
        "billed after",
        "billed again",
        "unauthorized charge",
        "unrecognized charge",
        "charged unexpectedly",
        "incorrect charge",
        "wrong amount",
        "money taken",
        "money deducted",
        "payment failed",
        "payment declined",
        "refund not received",
        "refund missing",
        "refund has not arrived"
    ]

    general_high = [
        "urgent",
        "immediately",
        "as soon as possible",
        "asap",
        "emergency",
        "critical",
        "need help now"
    ]

    if any(word in text for word in general_high):
        return "High"

    if category == "Account" and any(word in text for word in account_high):
        return "High"

    if category == "Technical" and any(word in text for word in technical_high):
        return "High"

    if category == "Billing" and any(word in text for word in billing_high):
        return "High"

    return "Normal"
```

```python
import csv
import os
import re
import threading
from datetime import datetime
from pathlib import Path

import gradio as gr

TICKET_FILE = Path("tickets.csv")
FEEDBACK_FILE = Path("flagged_tickets.csv")
TICKET_COLUMNS = ["ID", "Created", "Ticket", "Category", "Priority", "Team", "Status"]
FEEDBACK_COLUMNS = ["Ticket ID", "Reported", "Correct Category", "Note"]
STATUSES = ["Open", "In Progress", "Resolved"]
file_lock = threading.RLock()

department_map = {
    "Account": "Account Support",
    "Technical": "Technical Support",
    "Billing": "Billing Support"
}


def read_rows(path):
    if not path.exists() or path.stat().st_size == 0:
        return []
    with path.open("r", encoding="utf-8-sig", newline="") as file:
        return list(csv.DictReader(file))


def write_rows(path, columns, rows):
    temporary = path.with_name(path.name + ".tmp")
    with temporary.open("w", encoding="utf-8", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=columns)
        writer.writeheader()
        writer.writerows(rows)
    os.replace(temporary, path)


def dashboard(rows):
    counts = {name: sum(row["Category"] == name for row in rows)
              for name in department_map}
    high = sum(row["Priority"] == "High" for row in rows)
    open_count = sum(row["Status"] != "Resolved" for row in rows)
    return (
        f"### Dashboard\n"
        f"**Total:** {len(rows)}　 |　 **High priority:** {high}　 |　 "
        f"**Unresolved:** {open_count}\n\n"
        f"Account: {counts['Account']}　 ·　 "
        f"Technical: {counts['Technical']}　 ·　 "
        f"Billing: {counts['Billing']}"
    )


def history_view(rows):
    return [[row.get(column, "") for column in TICKET_COLUMNS]
            for row in reversed(rows)]


def current_view():
    with file_lock:
        rows = read_rows(TICKET_FILE)
    ids = [row["ID"] for row in reversed(rows)]
    return dashboard(rows), history_view(rows), gr.update(choices=ids)


def create_ticket(ticket_text):
    ticket_text = ticket_text.strip()
    if not ticket_text:
        stats, history, choices = current_view()
        return "Please enter a support ticket.", stats, history, choices

    category = str(model.predict(vectorizer.transform([ticket_text]))[0])
    priority = get_priority(ticket_text, category)
    with file_lock:
        rows = read_rows(TICKET_FILE)
        numbers = [int(match.group(1)) for row in rows
                   if (match := re.fullmatch(r"TKT-(\d+)", row["ID"]))]
        ticket_id = f"TKT-{max(numbers, default=0) + 1:04d}"
        record = {
            "ID": ticket_id,
            "Created": datetime.now().astimezone().strftime("%Y-%m-%d %H:%M %Z"),
            "Ticket": ticket_text,
            "Category": category,
            "Priority": priority,
            "Team": department_map[category],
            "Status": "Open"
        }
        rows.append(record)
        write_rows(TICKET_FILE, TICKET_COLUMNS, rows)

    result = (f"### {ticket_id} created\n"
              f"**Category:** {category}　 |　 **Priority:** {priority}\n\n"
              f"**Team:** {record['Team']}　 |　 **Status:** Open")
    stats, history, choices = current_view()
    return result, stats, history, gr.update(choices=choices["choices"], value=ticket_id)


def change_status(ticket_id, status):
    if not ticket_id:
        stats, history, choices = current_view()
        return "Choose a ticket first.", stats, history, choices
    with file_lock:
        rows = read_rows(TICKET_FILE)
        match = next((row for row in rows if row["ID"] == ticket_id), None)
        if match is None:
            message = "Ticket not found. Refresh the page and try again."
        else:
            match["Status"] = status
            write_rows(TICKET_FILE, TICKET_COLUMNS, rows)
            message = f"{ticket_id} is now {status}."
    stats, history, choices = current_view()
    return message, stats, history, gr.update(choices=choices["choices"], value=ticket_id if match else None)


def flag_ticket(ticket_id, correct_category, note):
    if not ticket_id:
        return "Choose a ticket first."
    with file_lock:
        rows = read_rows(TICKET_FILE)
        if not any(row["ID"] == ticket_id for row in rows):
            return "Ticket not found. Refresh the page and try again."
        feedback = read_rows(FEEDBACK_FILE)
        feedback.append({
            "Ticket ID": ticket_id,
            "Reported": datetime.now().astimezone().strftime("%Y-%m-%d %H:%M %Z"),
            "Correct Category": correct_category,
            "Note": note.strip()
        })
        write_rows(FEEDBACK_FILE, FEEDBACK_COLUMNS, feedback)
    return f"Feedback saved for {ticket_id}."


with gr.Blocks(title="Support Ticket Desk") as demo:
    gr.Markdown("# Support Ticket Desk\nClassify, route, and track sample support requests.")
    stats = gr.Markdown()

    with gr.Tab("New ticket"):
        ticket_text = gr.Textbox(
            label="Customer's message", lines=4,
            placeholder="Describe the customer's issue..."
        )
        gr.Examples(
            examples=[
                "I was charged twice for my subscription",
                "The application goes blank after opening",
                "I need to change the email linked to my account"
            ],
            inputs=ticket_text
        )
        submit = gr.Button("Create ticket", variant="primary")
        result = gr.Markdown()

    with gr.Tab("Ticket history"):
        history = gr.Dataframe(
            headers=TICKET_COLUMNS, datatype="str", interactive=False,
            label="Newest tickets first", wrap=True
        )
        with gr.Row():
            ticket_id = gr.Dropdown(label="Ticket ID", choices=[])
            status = gr.Dropdown(label="New status", choices=STATUSES, value="In Progress")
            save_status = gr.Button("Update status")
        status_result = gr.Markdown()

    with gr.Tab("Flag a ticket"):
        gr.Markdown("Save a correction to review later. This does not retrain the model.")
        flag_id = gr.Dropdown(label="Ticket ID", choices=[])
        correct_category = gr.Dropdown(
            label="Correct category", choices=list(department_map), value="Account"
        )
        note = gr.Textbox(label="Note (optional)", lines=2)
        flag = gr.Button("Save feedback")
        flag_result = gr.Markdown()

    def refresh():
        summary, tickets, choices = current_view()
        return summary, tickets, choices, gr.update(choices=choices["choices"])

    demo.load(refresh, outputs=[stats, history, ticket_id, flag_id])
    submit.click(
        create_ticket, inputs=ticket_text,
        outputs=[result, stats, history, ticket_id]
    ).then(
        lambda: gr.update(choices=current_view()[2]["choices"]), outputs=flag_id
    )
    save_status.click(
        change_status, inputs=[ticket_id, status],
        outputs=[status_result, stats, history, ticket_id]
    )
    flag.click(flag_ticket, inputs=[flag_id, correct_category, note], outputs=flag_result)

demo.launch()
```
